```python
def mesh_cut_and_separate(bigger = 'temp_meshNeuron.ply', smaller = 'temp_meshShaft.ply', output = 'temp_meshSpines.ply'):
    ms = pymeshlab.MeshSet()
    ms.load_new_mesh(bigger)
    ms.load_new_mesh(smaller)

    shaft = ms.mesh(1).vertex_matrix()
    
    ms.generate_boolean_difference(first_mesh=0,second_mesh=1)

    ms.save_current_mesh(str(output),save_vertex_normal=True)

    n_complete = ms.number_meshes()
    print(n_complete)

    ms.generate_splitting_by_connected_components()

    n_total = ms.number_meshes()
    print(n_total)

    means = []

    for i in range(n_complete,n_total):
        ms.set_current_mesh(new_curr_id = i)

        spine = ms.mesh(i).vertex_matrix()
        print(len(spine))
        spine_normal = ms.mesh(i).vertex_normal_matrix()
        print(len(spine_normal))
        vertices = []
        normals = []

        j = i + 1 - n_complete

        individual_file = open(f'DEV_saving_cutting_planes/indFile{j}.csv','a')

        for vertex_n in range(0,len(spine)):
            if np.any(shaft[:] == spine[vertex_n]):
                vertices.append(list(spine[vertex_n]))
                normals.append(list(spine_normal[vertex_n]))
                individual_file.write(f'{spine[vertex_n][0]},{spine[vertex_n][1]},{spine[vertex_n][2]} \n')
        individual_file.close()
        
        print('vertices len')
        print(len(vertices))
        print(normals)
        a = [item[0] for item in vertices]
        b = [item[1] for item in vertices]
        c = [item[2] for item in vertices]
        an = [item[0] for item in normals]
        bn = [item[1] for item in normals]
        cn = [item[2] for item in normals]

        means.append([np.median(a),np.median(b),np.median(c),np.median(an),np.median(bn),np.median(cn),vertices,normals])
        #print(np.mean(a),np.mean(b),np.mean(c))
        #print(np.mean(an),np.mean(bn),np.mean(cn))

        ms.save_current_mesh(f'temp_{j}.ply')

    

    print(means)
    with open('DEV_saving_cutting_planes/DEV_myfile.csv','w') as f:
        for sublist in means:
            for item in sublist:
                f.write(str(item) + ';')
            f.write('\n')
    n_spines = n_total - n_complete

    return([n_spines,0])

outcome = mesh_cut_and_separate()
```
